Variables are declared with the equal sign. As in C based languages, all expressions to the right of "=" are evaluated before being assigned to the variable on the left.

Python is both dynamically and strongly typed. This means that you don't need to explicitly declare the type of a variable in the code, as the interpreter assumes its type from the way it is used. This means you don't have to think too hard about typing, but errors are still thrown when you try to do something ill defined. 

```python
a = 42 
b = "The meaning of life."
a / b
```

All types in Python are *objects* (yes, even functions!). This is worth knowing, as many built-in types have strange interactions with certain operators. For instance, the multiplication operator between a string and integer does this: 

```python
a = 42 
b = "The meaning of life."
a * b
```

i.e. repeats the string *b*, *a* times. So, you still should be a bit careful with types.
